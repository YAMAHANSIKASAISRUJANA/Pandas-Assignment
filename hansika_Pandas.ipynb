### SECTION - D #Q1.A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing.

````python
#SECTION - D #Q1.Create a "student_assessment_dirty.csv"

import pandas as pd

df = pd.DataFrame({
    "Student_ID": ["S101", "S102", "S103", "S104", "S105", "S106", "S107",
                   "S108", "S102", "S109", "S110", "S111", "S112", "S113", "S114"],
    "Student_Name": ["Asha", "Ravi", "Neha", "Karan", "Meera", "Arjun", "Priya",
                     "Rahul", "Ravi", "Sneha", "Varun", "Diya", "Aman", "Nisha", "Kabir"],
    "Department": ["IT", "it", "I.T.", "HR", "HR", "Finance", "finance",
                   "IT", "it", "I.T.", "HR", "Finance", "IT", "HR", "Finance"],
    "Assessment": ["Python"] * 15,
     "Marks": ["85", "78", "92", "", "Absent", "0", "67", "75", "78",
               "NA", "88 marks", "91", "", "64", "105"]                              

})

df.to_csv("student_assessment_dirty.csv", index=False)
````

### SECTION - D #Q1.Data Cleaning of Student Assessment Dataset.

The dataset contains:
- Missing values in the Marks column.
- Duplicate student records.
- Inconsistent department names such as IT, it and I.T.
- Marks stored as text values(e.g., "88 marks", "Absent", "NA).

#### **Data Cleaning Sequence**:

1. Load the dataset into Pandas.

2. Standardize department names by converting variations such as IT, it and I.T. into a single format(IT).Similarly, convert finance and Finance into a consistent format(Finance).

3. Identify and remove duplicate student records.

4. Replace invalid or missing mark values such as Absent, NA and blank entries with NaN.

5. Convert the Marks column from text format to a numeric data type.

6. Investigate missing marks and distinguish absent students from students whose marks have not yet been entered.

7. Identify marks outside the valid range(0-100) and traet them as invalid values for further review or replacement.



### SECTION - D #Q1. STUDENT ASSESSMENT DATA CLEANING.

#### 1)STANDARDIZE DEPARTMENT NAMES: 

````python
#SECTION - D #Q1. 1)

df["Department"] = df["Department"].replace({
    "it": "IT",
    "I.T.": "IT",
    "finance": "Finance"
})

print(df["Department"])
````

#### **EXPLANATION**: Different department spellings of IT and Finance are converted into consistent names.

#### 2)HANDLE MISSING VALUES:

````python
#SECTION - D #Q1. 2)

print(df[df["Marks"].isna()][["Student_ID", "Student_Name", "Marks"]])
````

#### **EXPLANATION**: Missing marks are identified before deciding whether they should be removed or replaced.

#### 3)REMOVE DUPLICATE RECORDS:

````python
#SECTION - D #Q1. 3)

print("Duplicate records:", df.duplicated().sum())

df = df.drop_duplicates()

print("Records after removing duplicates:", len(df))
````

#### **EXPLANATION**: Duplicate records are removed to avoid counting of same student assessment more than once as it may lead to incorrect analysis and results.

#### 4)CONVERT MARKS TO NUMERIC:

````python
#SECTION - D #Q1. 4)

df["Marks"] = df["Marks"].replace("88 marks", "88")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df["Marks"])
````

#### **EXPLANATION**: Marks are converted into numeric values while Absent, NA and blank values become "NaN".

### **RISK OF DELETING MISSING VALUES**:
